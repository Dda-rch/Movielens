# MOVIELENS ОТЧЕТ

Давайте проведем анализ первых 1000 записей из movielens. Сначала посмотрим на сами фильмы (жанры, годы выпуска), затем на теги,
которыми их снабдили пользователи. После этого увидим оценки разных фильмов пользователями и пользователями разных фильмов. В конце подключим внешние данные из IMDB/TMDB, чтобы узнать про бюджеты, сборы и режиссёров.


```python
%timeit import movielens_analysis as mla
import movielens_analysis as mla
```

## 1. Загрузка данных

Создаём экземпляры всех четырёх классов модуля.

```python
%timeit -n1 -r1 mla.Movies('movies.csv')
movies = mla.Movies('movies.csv')
```

```python
%timeit -n1 -r1 tags = mla.Tags('tags.csv')
tags = mla.Tags('tags.csv')
```

```python
%timeit -n1 -r1 ratings = mla.Ratings('ratings.csv')
ratings = mla.Ratings('ratings.csv')
```

```python
%timeit -n1 -r1 links = mla.Links('links.csv', 'movies.csv')
links = mla.Links('links.csv', 'movies.csv')
```

## 2. Фильмы: жанры и годы выпуска

Смотрим, в каком году вышло больше всего фильмов из выборки и какие жанры встречаются чаще
всего.

```python
%timeit -n1 -r1 movies.dist_by_release()
movies.dist_by_release()
```

```python
%timeit -n1 -r1 movies.dist_by_genres()
movies.dist_by_genres()
```

Топ 10 фильмов с наибольшим числом жанров одновременно:

```python
%timeit -n1 -r1 movies.most_genres(10)
movies.most_genres(10)
```

```python
%timeit -n1 -r1 tags.tags_with('funny')
tags.tags_with('funny')
```

## 4. Оценки фильмов

Когда и какие оценки люди чаще всего ставили фильмам.

```python
%timeit -n1 -r1 ratings.movies.dist_by_year()
ratings.movies.dist_by_year()
```

```python
%timeit -n1 -r1 ratings.movies.dist_by_rating()
ratings.movies.dist_by_rating()
```

Какие фильмы оценивали чаще всего:

```python
%timeit -n1 -r1 ratings.movies.top_by_num_of_ratings(10)
ratings.movies.top_by_num_of_ratings(10)
```

`top_by_ratings` умеет считать среднее и медиану:

```python
%timeit -n1 -r1 ratings.movies.top_by_ratings(10)
ratings.movies.top_by_ratings(10)
```

```python
%timeit -n1 -r1 ratings.movies.top_by_ratings(10, metric='median')
ratings.movies.top_by_ratings(10, metric='median')
```

Самыми спорными стали этиф фильмы, зрительские оценки показали максимальный разброс:

```python
%timeit -n1 -r1 ratings.movies.top_controversial(10)
ratings.movies.top_controversial(10)
```

## 5. Оценки: со стороны пользователей

Класс `Ratings.Users` унаследован от `Ratings.Movies` и группирует те же данные, но по
`userId`. Смотрим на активность пользователей и на то, кто из них наименее последователен
в оценках.

Сколько оценок поставило сколько людей, и то же самое про среднюю оценку: (Ключи - количество оценок, а значения - количество людей)

```python
%timeit -n1 -r1 ratings.users.dist_by_num_of_ratings()
ratings.users.dist_by_num_of_ratings()
```

```python
%timeit -n1 -r1 ratings.users.dist_by_ratings()
ratings.users.dist_by_ratings()
```

Дисперсия оценок у конкретных пользователей, то есть тех кто чаще ставит крайние оценки

```python
%timeit -n1 -r1 ratings.users.top_controversial(5)
ratings.users.top_controversial(5)
```

## 6. Links: данные из TMDB

Класс `Links` объединяет `links.csv` и `movies.csv` и подтягивает бюджеты, сборы,
режиссёров и хронометраж из TMDB API. Данные кэшируются на диске, поэтому повторные
вызовы быстрые.

```python
%timeit -n1 -r1 links.get_imdb([1, 2, 3, 6, 10], ['Director', 'Budget', 'Runtime'])
links.get_imdb([1, 2, 3, 6, 10], ['Director', 'Budget', 'Runtime'])
```

Собрали с сайта tmdb информацию о 1000 фильмах в таком формате для дальнейшего анализа:

```python
%timeit -n1 -r1 links.get_movie_details(862)
links.get_movie_details(862)
```

Как много фильмов среди выборки снял конкретный режиссер:

```python
%timeit -n1 -r1 links.top_directors(5)
links.top_directors(5)
```

Самые дорогие фильмы:

```python
%timeit -n1 -r1 links.most_expensive(5)
links.most_expensive(5)
```

Самые прибыльные фильмы:

```python
%timeit -n1 -r1 links.most_profitable(5)
links.most_profitable(5)
```

Самые длительные фильмы:

```python
%timeit -n1 -r1 links.longest(5)
links.longest(5)
```

У каких фильмов самые дорогие минуты:

```python
%timeit -n1 -r1 links.top_cost_per_minute(5)
links.top_cost_per_minute(5)
```
